

*   Mean is the average value of the column.
*   Median is the middle value of sorted column.
*   Mode is the most frequent value
*   Standard deviation measures how the values spread out from the mean.
*   mean > median, the data is positively skewed because of a few very large values.
*   A large standard deviation means the values are spread over a wide range.


1.   Here the mean (10453.43) is much higher than the median (1450) in case of 'avg price', suggesting that a few very expensive products increase the average price.
2.   The Total Sales Value column has the highest standard deviation(50,535.07), indicating that sales values vary significantly across transactions.














Data Visualization

HISTOGRAMS

```python
df.hist(figsize=(12,8),bins=3,color='red')
plt.show()
```

```python
fig=plt.figure(figsize=(8,5))
sns.histplot(df['Volume'],bins=3,color='red',kde=True)
plt.xlabel('Volume')
plt.ylabel('Frequency')
plt.title('Histogram of Volume')
plt.show()
```

Inferences


*   Positively skewwd
*   Most transactions involve small quantities, while a few involve much larger quantities.



```python
fig=plt.figure(figsize=(8,5))
sns.histplot(df['Avg Price'],bins=4,color='red',kde=True)
plt.xlabel('Avg Price')
plt.ylabel('Frequency')
plt.title('Histogram of Avg Price')
plt.show()
```

Inferences


*   Highly positively skewed
*   There are few expensive products that increase the average price.
*   The histogram suggests the presence of high-value outliers because a small number of observations are located far from the main distribution.






```python
fig=plt.figure(figsize=(8,5))
sns.histplot(df['Total Sales Value'],bins=4,color='red',kde=True)
plt.xlabel('Total Sales Value')
plt.ylabel('Frequency')
plt.title('Histogram of Total Sales Value')
plt.show()
```

Inferences


*   Positively skewed.
*   Most transactions have moderate sales values, while a small number of transactions have exceptionally high sales values.



```python
fig=plt.figure(figsize=(8,5))
sns.histplot(df['Discount Rate (%)'],bins=4,color='red',kde=True)
plt.xlabel('Discount Rate (%)')
plt.ylabel('Frequency')
plt.title('Histogram of Discount Rate (%)')
plt.show()
```

Inferences


*   The distribution is slightly negatively (left) skewed, as most observations are concentrated at higher discount rates with a smaller number of lower discount rates.




```python
fig=plt.figure(figsize=(8,5))
sns.histplot(df['Discount Amount'],bins=4,color='red',kde=True)
plt.xlabel('Discount Amount')
plt.ylabel('Frequency')
plt.title('Histogram of Discount Amount')
plt.show()
```

Inferences


*   Positively skewed.
*   Large discounts are given only for a small number of transactions. The histogram suggests the presence of high-value outliers.



```python
fig=plt.figure(figsize=(8,5))
sns.histplot(df['Net Sales Value'],bins=4,color='red',kde=True)
plt.xlabel('Net Sales Value')
plt.ylabel('Frequency')
plt.title('Histogram of Net Sales Value')
plt.show()
```

Inferences


*   Positively skewed
*   Most net sales values are low, with a few much higher values.



BOXPLOTS

```python
df.plot(kind='box', subplots=True, layout=(3,2), figsize=(12,8),color='red')
plt.show()
plt.tight_layout()
plt.show()
```

Inferences
  
*   Avg Price, Total Sales Value, Discount Amount, and Net Sales Value contain several upper outliers, indicating unusually high transaction values.
*   The Volume column also has multiple upper outliers, while the Discount Rate (%) column contains a few lower outliers.
*   For Volume, Avg Price, Discount Rate (%), and Discount Amount, the IQR is narrow, indicating that the middle 50% of the values are concentrated within a small range.
*   For Net Sales Value,Total Sales Value, the IQR is relatively wide, indicating greater variation in the middle 50%.







BARCHARTS

```python
df["BU"].value_counts().plot(kind="bar")

plt.title("Business Unit Distribution")
plt.xlabel("Business Unit")
plt.ylabel("Count")
plt.show()
```

Inferences


*   The bar chart shows the frequency of transactions for each business unit.
*   Each Business unit has almost same transactions.



```python
df["Brand"].value_counts().plot(kind="bar")

plt.title("Brand Distribution")
plt.xlabel("Brand")
plt.ylabel("Count")
plt.show()
```

Inferences


*   The bar chart represents the number of transactions for each brand in the dataset.
*   Jeera brand has significantly higher transaction counts than others, indicating that they are more frequently sold.



```python
df["Day"].value_counts().plot(kind="bar")

plt.title("Day Distribution")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```




*   The bar chart shows the distribution of sales transactions across different days.
*   Thursday has higher transaction counts than others, indicating that sales activity varies depending on the day.



```python
df["Model"].value_counts().plot(kind="bar")

plt.title("Model Distribution")
plt.xlabel("Model")
plt.ylabel("Count")
plt.show()
```

Inferences


*   The bar chart shows the frequency of different product models.
*   All model has almost same frequency

The descriptive analytics and data visualization provided valuable insights into the Sales & Discounts dataset.
